```python
base_url = "https://fantasy.premierleague.com/api"
response = requests.get(f"{base_url}/bootstrap-static")
data = response.json()

if response.status_code == 200:
    data = response.json()
else:
    print(f"Request failed. Error: {response.text}")


Path("../bootrap-static_2026-09-03.json").write_text(json.dumps(data, indent=2))
```

```python
player_keys = {
    "id": int,
    "code": int,
    "opta_code": str,
    "web_name": str,
    "second_name": str,
    "first_name": str,
    "ep_next": float,
    "ep_this": float
}
```

```python
df = pl.from_dicts(players, schema=player_keys)
```

```python
df_ep = df.with_columns(
    (pl.col("ep_next") - pl.col("ep_this")).alias('ep_change')
).with_columns(
    pl.col('ep_change').sign().alias("sign")
)
```

```python
df_ep.group_by("sign").len()
```
